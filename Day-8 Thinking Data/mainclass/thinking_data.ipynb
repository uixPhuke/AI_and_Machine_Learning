# Cleaning and Structuring Data

- Data is inconsistent (rating is mixed: string and numeric values)
- Missing data (age is missing)
- Duplicate records may exist

```python
# Clean & Structured the data


def clean_data(data):
    text_to_num={"one":1,"two":2,"three":3,"four":4,"five":5}
    cleaned_data=[]
    unique_users=set()
    
    for user in data:
        #clean ratings 
        ## Unstructured rating
        raw_rating=str(user["rating"]).strip().lower()
        if(raw_rating in text_to_num):
            raw_rating=text_to_num[raw_rating]
        user["rating"]=raw_rating
        #print(raw_rating)

        ## missing age
        raw_age=user.get("age")
        if(raw_age==None):
            user["age"]=None
        

        ##duplicate data DE-DUPLICATION
        if(user["name"].strip() in unique_users):
            continue
        unique_users.add(user["name"])
        cleaned_data.append(user)

    return cleaned_data
        
        
        
```

```python
data=clean_data(data)
print(data)
```

# Assignment 2
## Getting meaningful insight from data
-  What's Average Rating ?
-  What % of users gave a poor rating (<3)?

```python
#Getting meaningful insight from data
def get_insights(data):
    
    #avg rating
    total_rating=0
    for user in data:
        total_rating += float(user["rating"])

    print(f"Average rating = {total_rating/len(data)}")

    #percentage of user with low rating
    poor_ratings=0
    for user in data:
        if(float(user["rating"]) < 3):
            poor_ratings+=1
    print(f"Percentage of user with low rating = {poor_ratings/len(data)*100}%")
            
```

```python
get_insights(data)
```

# Assignment 3
## Build a Product recommendation feature(Rule Based)
- If user rating >=4, recommend same brand products. (Apple)
- If user < 4, recommend different brand products. (Samsung)

```python
#Build the recommendation Feature  >=4 ->apple 
def get_recommendations(data):
    recommendations=[]

    for user in data:
        curr_recomm={}
        curr_recomm["name"]=user["name"]

        if(float(user["rating"])>=4):
            curr_recomm["brand"]="Apple"
        else:
            curr_recomm["brand"]="Samsung"
        recommendations.append(curr_recomm)
    return recommendations
    
            
            
            
```

```python
get_recommendations(data)
```
