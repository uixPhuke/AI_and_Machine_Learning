```python
import numpy as np
```

```python
arr=np.array([1,2,3,4,5])
print(arr,arr.dtype)
```

```python
arr=np.array([1, 2, 3.14, 4, 5])
print(arr,arr.dtype)
```

```python
arr=np.array([1, 2, 3.14, 4, 5, "phuke"])
print(arr,arr.dtype)
```

```python
arrComplex=np.array([9+5j])
arrComplex2=np.array([2+4j])


print(arrComplex,arrComplex.dtype)

print(arrComplex+arrComplex2)
print(arrComplex-arrComplex2)
```

```python
arr=np.array(["phuke",{1,2,3},3.14])
print(arr,arr.dtype)
```
