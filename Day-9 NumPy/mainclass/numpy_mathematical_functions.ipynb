```python
import numpy as np
```

# Aggregation

```python
arr=np.array([1,2,3,4,5])

print(np.sum(arr))
print(np.prod(arr))
print(np.max(arr))
print(np.min(arr))
print(np.mean(arr))
print(np.argmax(arr))
print(np.argmin(arr))
print(np.std(arr)) #sigma
print(np.var(arr)) #sigma^2
print(np.median(arr))
print(np.argmax(arr))
```

# Power Functons

```python
print(np.square(arr))
print(np.sqrt(arr))
print(np.pow(arr,3))
```

# Log Functions

```python
print(np.log(arr))
print(np.log10(arr))
print(np.log2(arr))
print(np.exp(arr))
```

# Rounding

```python
print(np.round(3.14))
print(np.ceil(3.14))
print(np.floor(3.14))
print(np.trunc(3.14))
```

```python
arr=np.array([1,-2,3,-4,5,3,5,1])
print(np.unique(arr))
print(np.sort(arr))
print(np.abs(arr))
```
