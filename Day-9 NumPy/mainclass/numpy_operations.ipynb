# RESHAPING

```python
import numpy as np
```

```python
arr=np.array([[1,2,3],[4,5,6]])
print(arr,arr.shape)

reshaped=arr.reshape((3,2))
print(reshaped,reshaped.shape)
```

```python
flattend=arr.flatten() # 2D=>1D
print(flattend,flattend.shape)
```

# INDEXING

```python
arr=np.array([1,2,3,4,5])

print(arr[0]) #1
print(arr[2]) #3
```

```python
arr2=np.array([[1,9,3],[4,5,6]])
print(arr2[0][2])
print(arr2[1][2])
```

# FANCY INDEXING

```python
arr=np.array([1,2,3,4,5])
index=[0,1,3]
print(arr[index])
```

# BOOLEAN INDEXING

```python
arr=np.array([1,2,3,4,5])

print(arr[arr>2])
print(arr[arr%2==0])
print(arr[arr%2!=0])
```

# SLICING

```python
arr=np.array([1,2,3,4,5])

print(arr[1:4]) #[start : end : step]
print(arr[1:]) #[start : end : step]
print(arr[:4]) #[start : end : step]
print(arr[:]) #[start : end : step]
print(arr[::2]) #[start : end : step]
```

# COPY VS VIEW

```python
nums=[1,2,3,4,5]
sub_list=nums[1:3]

print(sub_list)

sub_list[0]=77

print(sub_list)
print(nums)
```

```python
arr=np.array([1,2,3,4,5])
sub_arr=arr[1:3]

print(sub_arr)
sub_arr[0]=88

print(sub_arr)
print(arr)
```

```python
arr=np.array([1,2,3,4,5])
sub_arr=arr[1:3].copy()

print(sub_arr)
sub_arr[0]=88

print(sub_arr)
print(arr)
```
